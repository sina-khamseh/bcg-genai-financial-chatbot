```python
import pandas as pd

df = pd.read_csv('financial_data.csv')

df = df.sort_values(by=['Company', 'Fiscal_Year']).reset_index(drop=True)

metrics = [
    'Total_Revenue',
    'Net_Income',
    'Total_Assets',
    'Total_Liabilities',
    'Cash_Flow_Operating'
]

for col in metrics:
    df[f'{col}_Growth (%)'] = (df.groupby('Company')[col].pct_change() * 100).round(2)

df_filled = df.fillna(0)

df_filled.to_csv('processed_financial_data.csv', index=False)

display(df[['Company', 'Fiscal_Year', 'Total_Revenue', 'Total_Revenue_Growth (%)', 'Net_Income', 'Net_Income_Growth (%)']])
```

### Key Financial Insights & Summary (2024 - 2026)
1. **Revenue & Profit Growth Leader:** **Microsoft** demonstrated strong and consistent top-line and bottom-line expansion, with Total Revenue growing by **14.93%** in 2025 and **15.43%** in 2026, alongside a **22.52%** surge in Total Assets in 2026 driven by AI and cloud infrastructure investments.
2. **High Recovery & Volatility:** **Tesla** recorded significant Net Income growth rates (**18.09%** in 2025 and **16.98%** in 2026) while maintaining steady revenue expansion above **10%** YoY.
3. **Mature & Stable Cash Flow:** **Apple** maintained the largest overall revenue base ($435.8B in 2026) with steady single-digit revenue growth (**5.49%** to **5.65%**) and slightly reduced its Total Liabilities year-over-year, reflecting strong balance sheet management.
